```python
import numpy as np
```

```python
arr = np.random.randint(1,100,size=10)
print(arr)
sort = np.sort(arr)
print(sort)
arr.sort()
print("Array is sorted in the original one ",arr)
```

```python
arr = np.random.randint(1,100,size=10)
print(arr)
i=np.argsort(arr)
print(i)
sorted_=arr[i]
print("Sorted via fancy",sorted_)
np.sort(arr)
```

```python
np.random.seed(1)
ar =np.random.randint(19,220,size=(4,6))
print("Original Array",ar)
row_sorted= np.sort(ar,axis=0)
print("Row Sorted",row_sorted)
col_sorted= np.sort(ar,axis=1)
print("col sorted",col_sorted)
```

```python
np.random.seed(1)
ar =np.random.randint(1,22,size=15)
print(ar)
a1=np.partition(ar, 4)
print("3 smallest ",a1)
np.sort(ar)
```
